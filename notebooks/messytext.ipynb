# Imports

```python
import pandas as pd
```

# Dataset

```python
df = pd.read_csv(r"./datasets/messy_orders_cities.csv")
df
```

# Messy Text

```python
df["City"].value_counts()
```

```python
len(df["City"].value_counts())
```

```python
# Step 1 : strip
```

```python
df['City'] = df['City'].str.strip()
df
```

```python
len(df["City"].value_counts())
```

```python
# Step 2 : fix case
```

```python
df["City"] = df["City"].str.title()
```

```python
len(df["City"].value_counts())
```

```python
df["City"].value_counts()
```

```python
# Step 3 : replace
```

```python
df["City"] = df["City"].str.replace(r"[^\w\s]", "", regex=True)
df["City"] = df["City"].str.replace(r"\s+", " ", regex=True)
```

```python
len(df["City"].value_counts())
```

```python
df["City"].value_counts()
```

```python
# Step 4 : Standardize typos and abbreviations
```

```python
df["City"] = df['City'].replace({"Lndon": "London", "Ny": "New York",
                    "Berln" : "Berlin", "Pariss" : "Paris",
                    "NewYork" : "New York", "London Uk" : "London"})
```

```python
df["City"].value_counts()
```

```python
df["City"] = df['City'].replace("", None)
```

```python
df["City"].value_counts()
```

```python
df.dropna(subset=["City"], inplace=True)
```
