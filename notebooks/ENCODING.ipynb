# Imports

```python
import pandas as pd
from sklearn.preprocessing import OrdinalEncoder
```

# Dataset

```python
df = pd.read_csv(r"./datasets/sales_dataset.csv")
```

```python
df.head()
```

# Label Encoding

```python
enc = OrdinalEncoder(categories=[["small","medium","large"]])
df["size_encoded"] = (enc.fit_transform(df[["size"]]))
df
```

```python
df["encoded_way_2"] = df["size"].map({"small" : 0, "medium" : 1, "large" : 2})
df
```

# Hot Encoding

```python
dummies = pd.get_dummies(df[["color"]], dtype=int)
dummies
```

```python
df.drop(labels=["encoded_way_2"], axis=1, inplace=True)
```

```python
result = pd.concat([df, dummies], axis=1)
result
```
